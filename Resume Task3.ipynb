````python
from sklearn.model_selection import train_test_split

X = df["combined_text"]
y = df["fraudulent"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTrain distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTest distribution:")
print(y_test.value_counts(normalize=True) * 100)
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
````

````python
from sklearn.svm import LinearSVC

svm = LinearSVC(
    class_weight="balanced",
    C=1.0,
    random_state=42
)

svm.fit(X_train_tfidf, y_train)

y_pred = svm.predict(X_test_tfidf)
````

````python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

print("Classification Report:")
print(classification_report(y_test, y_pred, digits=4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# LinearSVC decision scores
y_score = svm.decision_function(X_test_tfidf)

print("\nROC-AUC:",
      roc_auc_score(y_test, y_score))

print("PR-AUC:",
      average_precision_score(y_test, y_score))
````

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from scipy.sparse import hstack, csr_matrix

categorical_cols = [
    "employment_type",
    "required_experience",
    "required_education",
    "industry",
    "function"
]

binary_cols = [
    "telecommuting",
    "has_company_logo",
    "has_questions"
]

# Use the same train/test indices
train_idx = X_train.index
test_idx = X_test.index

X_train_meta = df.loc[train_idx, categorical_cols + binary_cols].copy()
X_test_meta = df.loc[test_idx, categorical_cols + binary_cols].copy()

# One-hot encode categorical variables
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=True
)

X_train_cat = encoder.fit_transform(X_train_meta[categorical_cols])
X_test_cat = encoder.transform(X_test_meta[categorical_cols])

# Binary features
X_train_bin = csr_matrix(
    X_train_meta[binary_cols].astype(float).values
)

X_test_bin = csr_matrix(
    X_test_meta[binary_cols].astype(float).values
)

# Combine categorical + binary features
X_train_meta_encoded = hstack([
    X_train_cat,
    X_train_bin
]).tocsr()

X_test_meta_encoded = hstack([
    X_test_cat,
    X_test_bin
]).tocsr()

print("Metadata train shape:", X_train_meta_encoded.shape)
print("Metadata test shape:", X_test_meta_encoded.shape)
````

````python
X_train_combined = hstack([
    X_train_tfidf,
    X_train_meta_encoded
]).tocsr()

X_test_combined = hstack([
    X_test_tfidf,
    X_test_meta_encoded
]).tocsr()

print("Combined train shape:", X_train_combined.shape)
print("Combined test shape:", X_test_combined.shape)
````

````python
svm_combined = LinearSVC(
    class_weight="balanced",
    C=1.0,
    random_state=42
)

svm_combined.fit(X_train_combined, y_train)

y_pred_combined = svm_combined.predict(X_test_combined)
y_score_combined = svm_combined.decision_function(X_test_combined)
````

````python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

print("Classification Report:")
print(classification_report(y_test, y_pred_combined, digits=4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_combined))

print("\nROC-AUC:",
      roc_auc_score(y_test, y_score_combined))

print("PR-AUC:",
      average_precision_score(y_test, y_score_combined))
````

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

threshold_results = []

thresholds = np.arange(-0.50, 0.51, 0.01)

for fold, (train_idx_cv, val_idx_cv) in enumerate(
    skf.split(X_train_char_combined, y_train), 1
):

    print(f"\n===== Fold {fold} =====")

    X_cv_train = X_train_char_combined[train_idx_cv]
    X_cv_val = X_train_char_combined[val_idx_cv]

    y_cv_train = y_train.iloc[train_idx_cv]
    y_cv_val = y_train.iloc[val_idx_cv]

    model_cv = LinearSVC(
        class_weight="balanced",
        C=1.0,
        max_iter=5000,
        random_state=42
    )

    model_cv.fit(X_cv_train, y_cv_train)

    y_cv_score = model_cv.decision_function(X_cv_val)

    best_f1 = -1
    best_threshold = None
    best_precision = None
    best_recall = None

    for threshold in thresholds:

        y_cv_pred = (y_cv_score >= threshold).astype(int)

        precision = precision_score(
            y_cv_val,
            y_cv_pred,
            zero_division=0
        )

        recall = recall_score(
            y_cv_val,
            y_cv_pred,
            zero_division=0
        )

        f1 = f1_score(
            y_cv_val,
            y_cv_pred,
            zero_division=0
        )

        if f1 > best_f1:
            best_f1 = f1
            best_threshold = threshold
            best_precision = precision
            best_recall = recall

    threshold_results.append({
        "fold": fold,
        "best_threshold": best_threshold,
        "precision": best_precision,
        "recall": best_recall,
        "f1": best_f1
    })

    print(f"Best Threshold: {best_threshold:.2f}")
    print(f"Precision     : {best_precision:.4f}")
    print(f"Recall        : {best_recall:.4f}")
    print(f"F1            : {best_f1:.4f}")
````
